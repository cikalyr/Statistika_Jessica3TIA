```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import pearsonr, t
from google.colab import files
```

```python
file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("Data berhasil dibaca!")
print("Ukuran data:", df.shape)

df.head()
```

```python
print("Jumlah responden:", len(df))

print("Jumlah kolom:", len(df.columns))

print("\nNama seluruh kolom:")
print(df.columns.tolist())

print("\nJumlah data kosong:")
print(df.isnull().sum().sum())
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 2

butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))

butir.head()
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

print("Sel yang gagal dikonversi:",
      int(num.isna().sum().sum()))

num.head()
```

```python
import re
```

```python
dimensi = {}

for kolom in butir.columns:
    nama = re.match(r"^(.*?) \[", kolom).group(1)
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
from scipy import stats
```

```python
n = len(num)

t = stats.t.ppf(0.975, n - 2)

r_tabel = t / np.sqrt((n - 2) + t**2)

print(f"Jumlah responden (n) = {n}")
print(f"Derajat bebas (df) = {n-2}")
print(f"r tabel (α=0,05) = {r_tabel:.3f}")
```

```python
hasil_validitas = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)

        r_koreksi, _ = stats.pearsonr(
            x[kolom],
            total - x[kolom]
        )

        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid"
        })

tabel_validitas = pd.DataFrame(hasil_validitas)

display(tabel_validitas)
```

```python
def cronbach_alpha(x):
    k = x.shape[1]

    return k / (k - 1) * (
        1 - x.var(ddof=1).sum() /
        x.sum(axis=1).var(ddof=1)
    )


def kategori(a):
    return (
        "Sangat baik" if a >= 0.90
        else "Baik" if a >= 0.80
        else "Reliabel" if a >= 0.70
        else "Tidak reliabel"
    )
```

```python
hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():

    x = num[daftar]

    a = cronbach_alpha(x)

    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a)
    })

    for i, kolom in enumerate(daftar, 1):

        if len(daftar) > 2:

            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(
                    cronbach_alpha(x.drop(columns=kolom)), 3
                )
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)

tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

print("HASIL UJI RELIABILITAS")
display(tabel_reliabilitas)

print("HASIL ALPHA JIKA BUTIR DIHAPUS")
display(tabel_alpha_dihapus)
```

```python
display(
    tabel_alpha_dihapus.sort_values(
        by=["Dimensi", "Butir"]
    ).reset_index(drop=True)
)
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus,
    on=["Dimensi", "Butir"],
    how="left"
)

display(gabung)
```

```python
with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:

    gabung.to_excel(
        w,
        sheet_name="Validitas",
        index=False
    )

    tabel_reliabilitas.to_excel(
        w,
        sheet_name="Reliabilitas",
        index=False
    )

print("Hasil pengujian berhasil disimpan!")
```

```python
hasil_deskriptif = []

for nama, daftar in dimensi.items():

    total = num[daftar].sum(axis=1)

    hasil_deskriptif.append({
        "Dimensi": nama,
        "Mean": round(total.mean(), 3),
        "Standar Deviasi": round(total.std(), 3),
        "Skewness": round(total.skew(), 3),
        "Kurtosis": round(total.kurt(), 3)
    })

tabel_deskriptif = pd.DataFrame(hasil_deskriptif)

print("HASIL ANALISIS STATISTIK DESKRIPTIF")

display(tabel_deskriptif)
```

```python
for nama, daftar in dimensi.items():

    jawaban = num[daftar].stack()

    frekuensi = jawaban.value_counts().sort_index()

    plt.figure(figsize=(8, 5))

    plt.bar(
        frekuensi.index,
        frekuensi.values
    )

    plt.title(f"Komposisi Jawaban - {nama}")
    plt.xlabel("Skala Likert")
    plt.ylabel("Jumlah Jawaban")

    plt.xticks(
        [1, 2, 3, 4, 5],
        ["STS", "TS", "N", "S", "SS"]
    )

    plt.grid(axis="y", linestyle="--", alpha=0.5)

    plt.tight_layout()
    plt.show()
```
